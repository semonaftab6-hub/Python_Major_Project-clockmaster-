```python

from datetime import datetime
import time

# ==========================================
# CLOCK MASTER
# Smart Time Management System
# ==========================================

# Data Storage
alarms = []
study_records = []


# ==========================================
# 1. DIGITAL CLOCK
# ==========================================

def digital_clock():
    print("\n--- DIGITAL CLOCK ---")
    print("Press Ctrl + C to stop the clock.")

    try:
        while True:
            current_time = datetime.now()
            print(current_time.strftime("%d-%m-%Y %I:%M:%S %p"))
            time.sleep(1)

    except KeyboardInterrupt:
        print("\nDigital clock stopped.")


# ==========================================
# 2. ALARM CLOCK
# ==========================================
# Define the alarm clock function

def alarm_clock(alarm_time):

    print("\n--- ALARM CLOCK ---")

    try:
        # Validate alarm time
        alarm_time = datetime.strptime(alarm_time, "%H:%M").strftime("%H:%M")

        print("Alarm set for:", alarm_time)
        print("Waiting for the alarm...")
        print("Press Ctrl + C to cancel.")

        while True:

            current_time = datetime.now().strftime("%H:%M")

            print("Current Time:", datetime.now().strftime("%H:%M:%S"), end="\r")

            if current_time == alarm_time:

                print("\n\n⏰ WAKE UP! ALARM RINGING! ⏰")
                print("\a")
                break

            time.sleep(1)

    except ValueError:
        print("\nInvalid time format!")
        print("Please enter time in HH:MM format.")

    except KeyboardInterrupt:
        print("\nAlarm cancelled.")

# ==========================================
# 3. STOPWATCH
# ==========================================

def stopwatch():

    print("\n--- STOPWATCH ---")
    print("Stopwatch Started!")
    print("Press Enter to record a lap.")
    print("Enter S to stop the stopwatch.")

    start_time = time.time()
    lap_number = 1

    try:

        while True:

            user_input = input(
                "\nPress Enter for lap or S to stop: "
            )

            elapsed_time = time.time() - start_time

            minutes = int(elapsed_time // 60)
            seconds = int(elapsed_time % 60)
            milliseconds = int((elapsed_time % 1) * 1000)

            if user_input.lower() == "s":

                print("\n--- STOPWATCH STOPPED ---")
                print(
                    "Total Time:",
                    minutes, "minutes",
                    seconds, "seconds",
                    milliseconds, "milliseconds"
                )
                break

            elif user_input == "":

                print(
                    "Lap", lap_number, ":",
                    minutes, "min",
                    seconds, "sec",
                    milliseconds, "ms"
                )

                lap_number += 1

            else:

                print("Invalid input! Press Enter or S.")

    except KeyboardInterrupt:

        print("\nStopwatch cancelled by user.")




# ==========================================
# 4. COUNTDOWN TIMER
# ==========================================

def countdown_timer(total_minutes):

    total_seconds = total_minutes * 60

    print("\nCountdown started!")

    while total_seconds > 0:

        minutes = total_seconds // 60
        seconds = total_seconds % 60

        print(
            "Time Remaining:",
            minutes, "minutes", seconds, "seconds"
        )

        time.sleep(1)
        total_seconds -= 1

    print("\nTime is up!")


def custom_timer():
    print("\n--- COUNTDOWN TIMER ---")

    try:
        minutes = int(input("Enter countdown time in minutes: "))

        if minutes > 0:
            countdown_timer(minutes)

        else:
            print("Please enter a positive number.")

    except ValueError:
        print("Please enter a valid number.")
    except KeyboardInterrupt:
        print("\n timer cancelled.")


# ==========================================
# 5. SMART STUDY PLANNER
# ==========================================

def study_planner():
    print("\n--- SMART STUDY PLANNER ---")

    subject = input("Enter subject name: ")
    topic = input("Enter topic name: ")

    try:
        duration = int(input("Enter study duration in minutes: "))

        if duration > 0:

            print("\nYour study session is starting!")

            countdown_timer(duration)

            record = {
                "subject": subject,
                "topic": topic,
                "duration": duration,
                "date": datetime.now().strftime("%d-%m-%Y")
            }

            study_records.append(record)

            print("Study session completed and saved!")

        else:
            print("Please enter a positive duration.")

    except ValueError:
        print("Please enter a valid number.")

    except KeyboardInterrupt:
        print("\nSession cancelled by user.")
# ==========================================
# 6. STUDY HISTORY
# ==========================================

def study_history():
    print("\n--- STUDY HISTORY ---")

    if len(study_records) == 0:

        print("No study records available.")

    else:

        for i, record in enumerate(study_records, start=1):

            print("\nSession:", i)
            print("Subject:", record["subject"])
            print("Topic:", record["topic"])
            print("Duration:", record["duration"], "minutes")
            print("Date:", record["date"])


# ==========================================
# 7. PRODUCTIVITY STATISTICS
# ==========================================

def productivity_statistics():
    print("\n--- PRODUCTIVITY STATISTICS ---")

    total_minutes = 0

    for record in study_records:
        total_minutes += record["duration"]

    total_sessions = len(study_records)
    total_hours = total_minutes / 60

    print("Total Study Sessions:", total_sessions)
    print("Total Study Time:", total_minutes, "minutes")
    print("Total Study Hours:", round(total_hours, 2))


# ==========================================
# 8. DAILY PRODUCTIVITY SUMMARY
# ==========================================

def daily_productivity_summary():
    print("\n--- DAILY PRODUCTIVITY SUMMARY ---")

    today = datetime.now().strftime("%d-%m-%Y")

    daily_minutes = 0
    daily_sessions = 0

    for record in study_records:

        if record["date"] == today:

            daily_minutes += record["duration"]
            daily_sessions += 1

    print("Date:", today)
    print("Today's Study Sessions:", daily_sessions)
    print("Today's Study Time:", daily_minutes, "minutes")


# ==========================================
# 9. MAIN MENU
# ==========================================

def main():

    print("\n" + "=" * 40)
    print("          CLOCK MASTER")
    print("  Smart Time Management System")
    print("=" * 40)

    while True:

        print("\n========== MAIN MENU ==========")
        print("1. Digital Clock")
        print("2. Alarm Clock")
        print("3. Stopwatch")
        print("4. Countdown Timer")
        print("5. Smart Study Planner")
        print("6. Study History")
        print("7. Productivity Statistics")
        print("8. Daily Productivity Summary")
        print("9. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            digital_clock()

        elif choice == "2":
            alarm_time = input("Enter alarm time (HH:MM): ")
            alarm_clock(alarm_time)   

        elif choice == "3":
            stopwatch()

        elif choice == "4":
            custom_timer()

        elif choice == "5":
            study_planner()

        elif choice == "6":
            study_history()

        elif choice == "7":
            productivity_statistics()

        elif choice == "8":
            daily_productivity_summary()

        elif choice == "9":
            print("\nThank you for using CLOCK MASTER!")
            print("Have a productive day!")
            break

        else:
            print("Invalid choice! Please try again.")


# ==========================================
# START PROJECT
# ==========================================

main()
```
